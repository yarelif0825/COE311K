```python
import numpy as np

def jacobi(A, b, tol, max_iter=1000):
    x = np.zeros_like(b)
    for k in range(max_iter):
        x_new = np.array([
            (b[i] - sum(A[i][j]*x[j] for j in range (len(b)) if j != i)) / A[i][i] 
            for i in range(len(b))
        ])
        if np.linalg.norm(x_new - x, np.inf) < tol: 
            return x_new, k+1
        x = x_new
    return x, max_iter

def gauss_seidel(A, b, tol, max_iter=10000):
    x = np.zeros_like(b)
    for k in range(max_iter):
        x_new = np.copy(x)
        for i in range(len(b)):
            x_new[i] = (b[i] - sum(A[i][j]*x_new[j] for j in range(i)) - sum(A[i][j]*x[j] for j in range(i+1, len(b)))) / A[i][i]
        if np.linalg.norm(x_new - x, np.inf) < tol: 
            return x_new, k+1
        x = x_new
    return x, max_iter

A = np.array([[3, -0.1, -0.2],
              [0.1, 7, -0.3],
              [0.3, -0.2, 10]])
b = np.array([7.85, -19.3, 71.4])
tols = [1e-3, 1e-6, 1e-9]

print(" Tol   Jacobi iters   GS iters")
for tol in tols: 
    _, it_jacobi = jacobi(A, b, tol)
    _, it_gs = gauss_seidel(A, b, tol)
    print(f"{tol:<8}   {it_jacobi:^13}   {it_gs:^8}")
```
