```python
import numpy as np
import time
import matplotlib.pyplot as plt

def naive_mult(A, B):
    n = len(A)
    res = [[0]*n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                res[i][j] += A[i][k] * B[k][j]
    return res

sizes = [3 * (2**i) for i in range(8)]
naive_times = []
numpy_times = []

for n in sizes:
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)
    # Naive time
    start = time.time()
    naive_mult(A.tolist(), B.tolist())
    naive_times.append(time.time() - start)
    # Numpy time
    start = time.time()
    np.dot(A, B)
    numpy_times.append(time.time() - start)

plt.plot(sizes, naive_times, label='Naive', marker='o')
plt.plot(sizes, numpy_times, label='Numpy', marker='o')
plt.xlabel('Matrix Size')
plt.ylabel('Runtime (sec)')
plt.title('Naive vs Numpy Matrix Multiplication')
plt.legend()
plt.yscale('log')
plt.show()
```
