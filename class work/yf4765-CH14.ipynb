```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
a = 0
b = 2 * np.pi
N = 100
h = (b - a) / (N - 1)
x = np.linspace(a, b, N)
f = np.sin(x)

# True derivative
true_deriv = np.cos(x)

# Forward Difference
forward_diff = (np.sin(x + h) - np.sin(x)) / h

# Backward Difference
backward_diff = (np.sin(x) - np.sin(x - h)) / h

# Central Difference
central_diff = (np.sin(x + h) - np.sin(x - h)) / (2 * h)

# Plotting
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, true_deriv, label='cos(x)', color='black')
plt.title('True Derivative')
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(x, forward_diff, label='Forward Diff', color='blue')
plt.title('Forward Difference')
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(x, backward_diff, label='Backward Diff', color='green')
plt.title('Backward Difference')
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(x, central_diff, label='Central Diff', color='red')
plt.title('Central Difference')
plt.legend()

plt.tight_layout()
plt.show()
```
