```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2 * y

def y_true(x):
    return np.exp(-2 * x)

x0, x1 = 0, 2
h_values = np.arange(0.01, 1.01, 0.05)
errors = []

plt.figure(figsize=(10,6))

for h in h_values:
    N = int((x1 - x0) / h)
    x = np.linspace(x0, x1, N+1)
    y = np.zeros(N+1)
    y[0] = 1
    for i in range(N):
        y[i+1] = y[i] + h * f(x[i], y[i])
    y_exact = y_true(x)
    err = np.abs(y - y_exact)
    total_err = np.sum(err)
    errors.append(total_err)
    print(f"h={h:.2f}, total error={total_err:.6f}")
    plt.plot(x, y, label=f"h={h:.2f}")

plt.plot(np.linspace(x0, x1, 200), y_true(np.linspace(x0, x1, 200)), 'k--', label='True Solution')
plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler's Method for dy/dx = -2y")
plt.legend()
plt.show()

plt.figure()
plt.plot(h_values, errors, 'o-')
plt.xlabel('Step size (h)')
plt.ylabel('Total Error')
plt.title("Total Error vs Step Size")
plt.show()
```
