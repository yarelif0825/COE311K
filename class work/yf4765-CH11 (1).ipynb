```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

cs = CubicSpline(x, y, bc_type='natural')
x_interp = np.linspace(0, 5, 100)
y_spline = cs(x_interp)

coeffs = np.polyfit(x, y, 3)
y_poly = np.polyval(coeffs, x_interp)

rmse_poly = np.sqrt(np.mean((np.polyval(coeffs, x) - y) ** 2))
rmse_spline = np.sqrt(np.mean((cs(x) - y) ** 2))

plt.figure(figsize=(10, 6))
plt.plot(x, y, 'o', label='Original Data', markersize=8)
plt.plot(x_interp, y_spline, label='CubicSpline', linewidth=2)
plt.plot(x_interp, y_poly, '--', label='Polyfit (deg 3)', linewidth=2)
plt.xlabel('x')
plt.ylabel('y')
plt.title(f'CubicSpline vs Polyfit\nRMSE Spline: {rmse_spline:.3f}, Polyfit: {rmse_poly:.3f}')
plt.legend()
plt.grid(True)
plt.show()

print(f'RMSE at data points - CubicSpline: {rmse_spline:.4f}')
print(f'RMSE at data points - Polyfit deg 3: {rmse_poly:.4f}')
```
