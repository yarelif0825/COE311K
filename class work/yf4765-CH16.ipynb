```python
import numpy as np
import matplotlib.pyplot as plt

# PART 1: Compare different step sizes h

k = 1.0
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

x_start = 0
x_end = 10
y_0 = 1.0
v_0 = 0.0

plt.figure(figsize=(12, 6))

for h in h_values:
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(x_values)):
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]

        y_values[i] = y_n + h * v_n
        v_values[i] = v_n - h * k * y_n

    plt.plot(x_values, y_values, label=f'h = {h}')

# Solution
x_exact = np.linspace(x_start, x_end, 1000)
y_exact = np.cos(np.sqrt(k) * x_exact)

plt.plot(x_exact, y_exact, 'k--', linewidth=2, label='Exact solution')

plt.xlabel('x')
plt.ylabel('y(x)')
plt.title("Euler's Method for $y'' = -ky$ with Different Step Sizes")
plt.legend()
plt.grid(True)
plt.show()


# PART 2:  k = 10, h = 0.01

k = 10.0
h = 0.01

x_values = np.arange(x_start, x_end + h, h)
y_values = np.zeros(len(x_values))
v_values = np.zeros(len(x_values))

y_values[0] = y_0
v_values[0] = v_0

for i in range(1, len(x_values)):
    y_n = y_values[i - 1]
    v_n = v_values[i - 1]

    y_values[i] = y_n + h * v_n
    v_values[i] = v_n - h * k * y_n

# solution for k = 10
x_exact = np.linspace(x_start, x_end, 1000)
y_exact = np.cos(np.sqrt(k) * x_exact)

plt.figure(figsize=(10, 5))
plt.plot(x_values, y_values, label='Euler approximation')
plt.plot(x_exact, y_exact, 'k--', linewidth=2, label='Exact solution')

plt.xlabel('x')
plt.ylabel('y(x)')
plt.title("Euler's Method with k = 10 and h = 0.01")
plt.legend()
plt.grid(True)
plt.show()


# PART 3: Questions

print("QUESTIONS") 

print("1. When does it become unstable?")
print("For any h>0  value, Euler’s Method is theoretically unstable for this oscillatory problem.")
print("However, instability is clearly visible for larger step sizes, especially around h >= 0.1.\n")

print("2. Is it becoming unstable?")
print("Yes, it is becoming unstable. The numerical solution shows increasing amplitude over time instead of remaining bounded like the exact solution.\n")

print("3. Effect of changing k:")
print("As k increases, the system oscillates faster, or at a higher frequency.")
print("This makes Euler’s Method less accurate and causes instability to appear more quickly.\n")

print("4. k = 10, h = 0.01:")
print("The solution appears reasonable over a short interval, but still slowly becomes unstable and deviates from the exact solution.\n")

print("5. Can Euler’s Method work with larger h?")
print("No, Euler's method will not work with larger h. Larger step sizes cause significant error and faster instability, making Euler’s Method unreliable for this problem.\n")
```
